# FED4FED: Statistical Algorithms Validation
This notebook verifies Algorithm 1 (Graybill-Deal / T1), Algorithm 2 (Welch-type F test), and Algorithm 3 (Confidence Interval) using analytical formulas and controlled Monte Carlo synthetic distributions.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import numpy as np
from src.phase2_fed4fed.alg1_graybill_deal import run_algorithm_1
from src.phase2_fed4fed.alg2_welch import run_algorithm_2
from src.phase2_fed4fed.alg3_confidence import run_algorithm_3
from src.phase2_fed4fed.fairness import evaluate_fairness
```

## 1. Homogeneous Case (Fair Scenario Verification)

```python
np.random.seed(42)
K = 30
# Common population performance around 0.85 with small estimation variance
means_fair = np.random.normal(0.85, 0.008, size=K)
variances_fair = np.random.uniform(0.0004, 0.0012, size=K)

res_fair = evaluate_fairness(means_fair, variances_fair, significance_level=0.05)
print("Fair Scenario Verdict:", res_fair['overall_verdict'])
print("Alg 1 p-value:", res_fair['alg1_result']['p_value'])
print("Alg 2 p-value:", res_fair['alg2_result']['p_value'])
print("Algorithm 3 Permitted:", res_fair['confidence_interval']['is_permitted'])
```

## 2. Heterogeneous Case (Unfair Scenario Verification)

```python
# Clients split between high and low performing groups
means_unfair = np.concatenate([np.random.normal(0.92, 0.005, size=K//2), np.random.normal(0.68, 0.005, size=K//2)])
variances_unfair = np.random.uniform(0.0004, 0.0012, size=K)

res_unfair = evaluate_fairness(means_unfair, variances_unfair, significance_level=0.05)
print("Unfair Scenario Verdict:", res_unfair['overall_verdict'])
print("Alg 1 p-value:", res_unfair['alg1_result']['p_value'])
print("Alg 2 p-value:", res_unfair['alg2_result']['p_value'])
print("Algorithm 3 Status:", res_unfair['confidence_interval']['status'])
```
